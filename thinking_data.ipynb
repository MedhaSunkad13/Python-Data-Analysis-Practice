```python
import json
```

```python
#Load the data
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)

    return data
```

```python
data = load_data("store_data.json")
print(data)
print(type(data))
```

```python
#Clean and Structure the data
def clean_data(data):
    text_to_num = {
        "one": 1,
        "two": 2,
        "three": 3,
        "four": 4,
        "five": 5
    }

    cleaned_data = []
    unique_users = set()

    for user in data:
        #Make corrections of data inconsistency
        raw_rating = user["rating"]

        if isinstance(raw_rating, str):
            raw_rating = raw_rating.strip().lower()

            if raw_rating in text_to_num:
                user["rating"] = text_to_num[raw_rating]
                
        #Handle missing data
        raw_age = user.get('age')
        if raw_age == None:
            user["age"] = None

        #Deduplication
        if user["name"].strip() in unique_users:
            continue

        unique_users.add(user["name"])
        cleaned_data.append(user)

    return cleaned_data
```

```python
new_data = clean_data(data)
```

```python
#Get meaningful insights from data

def get_insights(data):

    #avg rating
    tot_rating = 0

    for user in data:
        tot_rating += float(user['rating'])

    print(f"Avg Rating = {tot_rating / len(new_data)}")

    #Percentage of user's with poor rating
    poor_rating = 0
    for user in data:
        
        if float(user['rating']) < 3:
            poor_rating += 1

    print(f"Percentage of users with poor rating : {poor_rating / len(new_data) * 100}%")
```

```python
get_insights(new_data)
```

```python
#Recommending products for users which has good ratings or reviews
#Build recommendation feature

def get_recommendations(data):
    recommendations = []

    for user in data:
        curr_recomm = {}
        curr_recomm["name"] = user["name"]
        
        if(float(user["rating"]) >= 4):
           curr_recomm["brand"] = "Apple"
        else:
            curr_recomm["brand"] = "Samsung"

        recommendations.append(curr_recomm)

    return recommendations
```

```python
get_recommendations(new_data)
```
